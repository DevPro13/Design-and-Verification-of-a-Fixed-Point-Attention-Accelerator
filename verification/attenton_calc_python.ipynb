```python
import math
import numpy as np
import random
import time
import re
```

```python
MySentence="Nepal does not have direct access to the sea because it is a landlocked country."
```

# Creating Embeddings
--> Considering embedding vector size 4

```python
embedding_dim=3
low=-0.1
high=0.3
random.seed(time.time())
```

```python
x = round(random.uniform(low,high),3)
print(x)
```

```python
def tokenize(text):
    return re.findall(r"\w+|[^\w\s]", text)
Tokens=tokenize(MySentence)
Tokens
```

# Calculating Embeddings of each tokens

```python
embeddings=[[round(random.uniform(low,high),3) for _ in range(embedding_dim)] for _ in Tokens]
embeddings_vector_array=np.array(embeddings)
embeddings_vector_array
```

```python
# Save txt
np.savetxt("token_embedding.txt",embeddings_vector_array,fmt="%.3f")
```

# Weight Matrices
* Learnable parameters that transform vectors having dimention embedding_dim x embedding_dim

```python
weightQ=np.array([
    [round(random.uniform(-1,2),2) for _ in range(embedding_dim)] for _ in range(embedding_dim)
])
weightK=np.array([
    [round(random.uniform(-1,2),2) for _ in range(embedding_dim)] for _ in range(embedding_dim)
])
weightV=np.array([
    [round(random.uniform(-1,2),2) for _ in range(embedding_dim)] for _ in range(embedding_dim)
])
print(weightQ)
print(weightK)
print(weightV)
```

```python
#save weights
np.savetxt("weight_q.txt",weightQ,fmt="%.3f")
np.savetxt("weight_k.txt",weightK,fmt="%.3f")
np.savetxt("weight_v.txt",weightV,fmt="%.3f")
```

# Calculating Query, Key And Values Vectors

```python
Q=embeddings_vector_array@weightQ
K=embeddings_vector_array@weightK
V=embeddings_vector_array@weightV
Q
```

# Calculating raw attention score

```python
Attention_score=Q@ K.T
Attention_score
```

# Scaling the score with key vector dimension

```python
Attention_score_scaled=Attention_score/np.sqrt(K.shape[0])
Attention_score_scaled
```

# Applying Softmax to calculate attention weights-->Probabilities

```python
type(Attention_score_scaled)
print(type(Attention_score_scaled))
print(np.shape(Attention_score_scaled))
```

```python
def softmax(xi):
    exp_x = np.exp(xi - np.max(xi, axis=1, keepdims=True))
    return exp_x / np.sum(exp_x, axis=1, keepdims=True)

attention_weights=softmax(Attention_score_scaled)
attention_weights
```

# Finally Calculating Context Vector

```python
context_vector=attention_weights@V
context_vector
```

```python
#save contextual vector
np.savetxt("output_context_vector.txt",context_vector,fmt="%.4f")
```
